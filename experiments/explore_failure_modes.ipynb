```python
import os
import json
import pandas as pd
from pathlib import Path
from typing import Dict, List, Set
import warnings
warnings.filterwarnings('ignore')

def extract_task_identifiers_from_path(model_trajectory_path: str) -> Dict[str, Set[str]]:
    """
    Extract unique task identifiers for each task type from a given path.
    
    Args:
        task_types_path: Path to directory containing task type folders
        
    Returns:
        Dictionary mapping task types to sets of task identifiers
    """
    model_trajectory_dir = Path(model_trajectory_path)
    task_identifiers = {}
    
    for task_type_dir in model_trajectory_dir.iterdir():
        if task_type_dir.is_dir() and not task_type_dir.name.startswith('.'):
            task_type = task_type_dir.name
            task_ids = set()
            
            for task_dir in task_type_dir.iterdir():
                if task_dir.is_dir():
                    # Check if result.txt exists and read success/failure
                    result_file = task_dir / "result.txt"
                    if result_file.exists():
                        try:
                            with open(result_file, 'r') as f:
                                result = f.read().strip()
                                if int(float(result)) == 0 or int(float(result)) == 1:
                                    task_ids.add(task_dir.name)
                        except Exception as e:
                            print(f"Error reading {result_file}: {e}")
            
            if task_ids:
                task_identifiers[task_type] = task_ids
    
    return task_identifiers

def load_task_configs(task_identifiers: Dict[str, Set[str]], examples_path: str) -> pd.DataFrame:
    """
    Load task configuration JSON files into a pandas DataFrame.
    
    Args:
        task_identifiers: Dictionary mapping task types to sets of task identifiers
        examples_path: Path to examples directory containing JSON configs
        
    Returns:
        DataFrame with task configurations
    """
    configs = []
    examples_dir = Path(examples_path)
    
    for task_type, task_ids in task_identifiers.items():
        task_type_path = examples_dir / task_type
        
        if not task_type_path.exists():
            print(f"Warning: Task type directory {task_type} not found in examples")
            continue
            
        for task_id in task_ids:
            config_file = task_type_path / f"{task_id}.json"
            
            if config_file.exists():
                try:
                    with open(config_file, 'r') as f:
                        config = json.load(f)
                        config['task_type'] = task_type
                        config['task_id'] = task_id
                        configs.append(config)
                except Exception as e:
                    print(f"Error loading {config_file}: {e}")
            else:
                print(f"Warning: Config file {config_file} not found")
    
    return pd.DataFrame(configs)

def get_task_result(task_types_path: str, task_type: str, task_id: str) -> int:
    """Get success/failure result for a specific task."""
    result_file = Path(task_types_path) / task_type / task_id / "result.txt"
    print(f'getting result from {result_file}')
    if result_file.exists():
        try:
            with open(result_file, 'r') as f:
                return int(float(f.read().strip()))
        except:
            return None
    return None

def analyze_trajectories(model_trajectory_path: str, examples_path: str, model_name: str = None) -> pd.DataFrame:
    """
    Analyze trajectories from a specific model's task types directory.
    
    Args:
        task_types_path: Path to directory containing task type folders
        examples_path: Path to examples directory containing JSON configs
        model_name: Optional model name to add to DataFrame
        
    Returns:
        DataFrame with task configurations and performance data
    """
    print(f"Analyzing trajectories from: {model_trajectory_path}")
    
    # Extract task identifiers
    task_identifiers = extract_task_identifiers_from_path(model_trajectory_path)
    
    if not task_identifiers:
        print(f"No task identifiers found in {model_trajectory_path}")
        return pd.DataFrame()
    
    # Load task configs
    df = load_task_configs(task_identifiers, examples_path)
    
    if df.empty:
        print(f"No task configs loaded from {examples_path}")
        return df
    
    # Add model information if provided
    if model_name:
        df['model'] = model_name
    
    # Add success/failure information
    df['success'] = df.apply(lambda row: get_task_result(model_trajectory_path, row['task_type'], row['task_id']), axis=1)
    
    return df

# Usage example:
if __name__ == "__main__":
    # Example usage for a specific model
    model_trajectory_base_folder = "/Users/lockewang/FIG/software-control/external_data/osworld-verified/"
    examples_path = "/Users/lockewang/FIG/software-control/src/OSWorld/evaluation_examples/examples"
    model_paths = [
        "uitars15-7b-100step",
        # "uitars15-7b-100step-1",
        # 'jedi-7b-o3-100steps',
        # 'o3_gta1_100steps/o3_gta1_100steps',
        # 'opencua_agent-opencua_32b-cot_l2-action_history-3image-Ubuntu-100steps/turn_1',
        # 'opencua_agent-opencua_32b-cot_l2-action_history-3image-Ubuntu-100steps/turn_2',
        # 'opencua_agent-opencua_32b-cot_l2-action_history-3image-Ubuntu-100steps/turn_3',
        # 'mobile-agent-v3-gui-owl-7b',
        # 'mobileagent_v3',
        # "autoglm_50steps",
        ]
    
    # all_dfs = {}
    for model_path in model_paths:
        model_trajectory_path = model_trajectory_base_folder + model_path
        print(f"Analyzing {model_trajectory_path}")
        df = analyze_trajectories(model_trajectory_path, examples_path, model_trajectory_path.split("/")[-1])
        
        if not df.empty:
            print(f"\nDataset shape: {df.shape}")
            print(f"Columns: {list(df.columns)}")
            
            # Basic exploratory analysis
            print("\n=== Basic Statistics ===")
            print(f"Total tasks: {len(df)}")
            print(f"Unique task types: {df['task_type'].nunique()}")
            print(f"Task types: {df['task_type'].value_counts().to_dict()}")
            
            # Success rate analysis
            if 'success' in df.columns:
                print("\n=== Success Rate Analysis ===")
                success_by_type = df.groupby('task_type')['success'].agg(['count', 'sum', 'mean']).round(3)
                success_by_type.columns = ['total_tasks', 'successful_tasks', 'success_rate']
                print(success_by_type)
            
            # Display first few rows
            print(f"\n=== Sample Data ===")
            print(df.head())
            # all_dfs[model_path] = df
        else:
            print("No data loaded")
```

```python
from IPython.display import Video, display
from pathlib import Path
import cv2
import glob

def load_trajectory_video(task_id: str, snapshot: str, model_dir: str, 
                         base_path: str = "/Users/lockewang/FIG/software-control/external_data/osworld-verified") -> Video:
    """
    Load and display MP4 video for a specific task trajectory.
    
    Args:
        task_id: Unique task identifier
        snapshot: Task type/snapshot name (e.g., 'chrome', 'vs_code')
        model_dir: Model directory name
        base_path: Base path to osworld-verified folder

    Returns:
        IPython Video object for display in Jupyter
    """
    video_base_path = Path(base_path) / model_dir / snapshot / task_id
    video_path = video_base_path / "recording.mp4"
    print(f'loading video from {video_path}')

    if not video_path.exists():
        print(f'video_path does not exist: {video_path}')
        png_file_list = sorted(glob.glob(str(video_base_path / "*.png")))
        print(f'found {len(png_file_list)} png files in {str(video_base_path)}')

        if not png_file_list:
            print(f"No PNG files found in {str(video_base_path)}. Cannot create video.")
            return None  # Or raise an appropriate error

        # Load first image to get consistent size
        frame = cv2.imread(png_file_list[0])
        if frame is None:
            print(f"Error reading first frame: {png_file_list[0]}")
            return None

        height, width, layers = frame.shape
        size = (width, height)

        # Use H.264 codec if available, fallback to mp4v
        fourcc = cv2.VideoWriter_fourcc(*'H264')
        video = cv2.VideoWriter(str(video_path), fourcc, 10, size)

        for image_path in png_file_list:
            img = cv2.imread(image_path)
            if img is None:
                print(f"⚠️ Skipping unreadable image: {image_path}")
                continue
            if img.shape[1] != width or img.shape[0] != height:
                print(f"⚠️ Resizing {image_path} to {size}")
                img = cv2.resize(img, size)
            video.write(img)

        video.release()
        cv2.destroyAllWindows()

    return Video(str(video_path), width=1600, height=1200)


def load_video_from_row(row: pd.Series) -> Video:
    """Load video from a DataFrame row containing task_id and snapshot columns."""
    return load_trajectory_video(row['task_id'], row['snapshot'], row['model'])

filtered_df = df[(df['task_type'] == 'chrome') & (df['success'] == 0) & (df['evaluator'] != "'func': 'infeasible'")]

print(f'found {len(filtered_df)} tasks')

from pprint import pprint

# Usage with DataFrame:
for i in range(5):
    print(f'Task: {filtered_df.iloc[i]["instruction"]}')
    print(f'''
    task_id: {filtered_df.iloc[i]["task_id"]}
    model: {filtered_df.iloc[i]["model"]}
    instruction: {filtered_df.iloc[i]["instruction"]}
    success: {filtered_df.iloc[i]["success"]}
    related_apps: {filtered_df.iloc[i]["related_apps"]}
    ''')
    pprint(filtered_df.iloc[i]["evaluator"])
    print('\n')

    video = load_video_from_row(filtered_df.iloc[i])

    display(video)
```
