```python
import os
import json
from pprint import pprint

trajectories_dir = "/Users/lockewang/FIG/software-control/trajectories"
folders = [f for f in os.listdir(trajectories_dir) if os.path.isdir(os.path.join(trajectories_dir, f))]

total_runs = 0
total_completed = 0
completed_seed_trajectory_ids = []

for folder in folders:
    run_folders = [f for f in os.listdir(os.path.join(trajectories_dir, folder)) if "run_" in f]
    newest_run_folder = max(run_folders, key=lambda x: os.path.getmtime(os.path.join(trajectories_dir, folder, x)))
    scenario_folders = [f for f in os.listdir(os.path.join(trajectories_dir, folder, newest_run_folder)) if os.path.isdir(os.path.join(trajectories_dir, folder, newest_run_folder, f))]
    scenario_folder = os.path.join(trajectories_dir, folder, newest_run_folder, scenario_folders[0])
    traj_file = os.path.join(scenario_folder, "traj.jsonl")
    try:
        with open(traj_file, "r") as f:
            traj = [json.loads(line) for line in f]
    except Exception as e:
        continue
        
    if traj[-1].get('completion_status'):
        # pprint(traj[-1]['completion_status'])
        total_runs += 1
        if traj[-1]['completion_status']['completed_all_steps'] == True:
            total_completed += 1
            completed_seed_trajectory_ids.append(folder)

print(f"Total runs: {total_runs}, Total completed: {total_completed}")
print("Completed seed trajectory ids: ")
pprint(completed_seed_trajectory_ids)
```
