```python
import nest_asyncio
from playwright.async_api import async_playwright
import json
import asyncio

# Apply nest_asyncio to allow nested event loops in Jupyter
nest_asyncio.apply()

async def highlight_element_by_backend_node_id(page, backend_node_id, color='red'):
    """
    Use CDP (Chrome DevTools Protocol) to resolve backend node ID and get bounding box
    """
    try:
        print(f"🔧 Using CDP to resolve backend_node_id: {backend_node_id}")
        
        # Create CDP session
        cdp_session = await page.context.new_cdp_session(page)
        
        # Step 1: Resolve backend node ID to DOM node object
        resolve_result = await cdp_session.send("DOM.resolveNode", {
            "backendNodeId": int(backend_node_id)
        })
        
        if "object" not in resolve_result:
            print(f"❌ Could not resolve backend_node_id {backend_node_id}")
            return None
            
        object_id = resolve_result["object"]["objectId"]
        print(f"✅ Resolved to object ID: {object_id}")
        
        # Step 2: Get the bounding box using CDP
        try:
            box_result = await cdp_session.send("DOM.getBoxModel", {
                "objectId": object_id
            })
            
            if "model" in box_result:
                # Extract coordinates from the box model
                # content_quad contains [x1,y1, x2,y2, x3,y3, x4,y4] for the content box
                content_quad = box_result["model"]["content"]
                
                # Calculate bounding rectangle from quad points
                x_coords = [content_quad[i] for i in range(0, len(content_quad), 2)]
                y_coords = [content_quad[i] for i in range(1, len(content_quad), 2)]
                
                x = min(x_coords)
                y = min(y_coords)
                width = max(x_coords) - min(x_coords)
                height = max(y_coords) - min(y_coords)
                
                location = {
                    'x': x,
                    'y': y,
                    'width': width,
                    'height': height,
                    'center_x': x + width / 2,
                    'center_y': y + height / 2
                }
                
                print(f"📦 CDP Bounding Box:")
                print(f"  📍 Position: ({location['x']:.1f}, {location['y']:.1f})")
                print(f"  📏 Size: {location['width']:.1f} x {location['height']:.1f}")
                print(f"  🎯 Center: ({location['center_x']:.1f}, {location['center_y']:.1f})")
                
                # Step 3: Highlight the element using CDP
                await cdp_session.send("Runtime.callFunctionOn", {
                    "functionDeclaration": f"""
                    function() {{
                        this.style.outline = '3px solid {color}';
                        this.style.outlineOffset = '2px';
                        this.style.backgroundColor = 'rgba({255 if color == 'red' else 0}, {0 if color == 'red' else 0}, {255 if color == 'blue' else 0}, 0.1)';
                        this.scrollIntoView({{behavior: 'smooth', block: 'center'}});
                        return true;
                    }}
                    """,
                    "objectId": object_id
                })
                
                print(f"✅ Successfully highlighted element via CDP")
                return location
                
        except Exception as box_error:
            print(f"⚠️ Could not get box model: {box_error}")
            
            # Fallback: Try to get bounding client rect via Runtime.callFunctionOn
            try:
                rect_result = await cdp_session.send("Runtime.callFunctionOn", {
                    "functionDeclaration": f"""
                    function() {{
                        const rect = this.getBoundingClientRect();
                        
                        // Highlight while we're at it
                        this.style.outline = '3px solid {color}';
                        this.style.outlineOffset = '2px';
                        this.style.backgroundColor = 'rgba({255 if color == 'red' else 0}, {0 if color == 'red' else 0}, {255 if color == 'blue' else 0}, 0.1)';
                        this.scrollIntoView({{behavior: 'smooth', block: 'center'}});
                        
                        return {{
                            x: rect.left,
                            y: rect.top,
                            width: rect.width,
                            height: rect.height,
                            center_x: rect.left + rect.width / 2,
                            center_y: rect.top + rect.height / 2
                        }};
                    }}
                    """,
                    "objectId": object_id,
                    "returnByValue": True
                })
                
                if rect_result.get("result", {}).get("value"):
                    location = rect_result["result"]["value"]
                    print(f"📦 Fallback getBoundingClientRect:")
                    print(f"  📍 Position: ({location['x']:.1f}, {location['y']:.1f})")
                    print(f"  📏 Size: {location['width']:.1f} x {location['height']:.1f}")
                    print(f"  🎯 Center: ({location['center_x']:.1f}, {location['center_y']:.1f})")
                    return location
                    
            except Exception as rect_error:
                print(f"⚠️ Fallback getBoundingClientRect also failed: {rect_error}")
        
        await cdp_session.detach()
        
    except Exception as e:
        print(f"❌ CDP method failed for backend_node_id {backend_node_id}: {e}")
        
        # Final fallback: Use the old selector-based method
        print("🔄 Falling back to selector-based method...")
        return await fallback_selector_method(page, backend_node_id, color)
    
    return None

async def fallback_selector_method(page, backend_node_id, color='red'):
    """
    Fallback method using selectors (your original approach)
    """
    try:
        fallback_script = f"""
        () => {{
            let element = null;
            
            // Try by id if the backend_node_id might correspond to an id
            element = document.getElementById('{backend_node_id}');
            if (element) {{
                console.log('Found by ID');
            }} else {{
                // Try by data attributes
                const selectors = [
                    '[data-backend-node-id="{backend_node_id}"]',
                    '[backend_node_id="{backend_node_id}"]',
                    '[data-testid*="{backend_node_id}"]',
                    '[data-node-id="{backend_node_id}"]'
                ];
                
                for (const selector of selectors) {{
                    element = document.querySelector(selector);
                    if (element) {{
                        console.log('Found by selector:', selector);
                        break;
                    }}
                }}
            }}
            
            if (element) {{
                element.style.outline = '3px solid {color}';
                element.style.outlineOffset = '2px';
                element.style.backgroundColor = 'rgba({255 if color == 'red' else 0}, {0 if color == 'red' else 0}, {255 if color == 'blue' else 0}, 0.1)';
                element.scrollIntoView({{behavior: 'smooth', block: 'center'}});
                
                const rect = element.getBoundingClientRect();
                return {{
                    x: rect.left,
                    y: rect.top,
                    width: rect.width,
                    height: rect.height,
                    center_x: rect.left + rect.width / 2,
                    center_y: rect.top + rect.height / 2
                }};
            }}
            
            return null;
        }}
        """
        
        result = await page.evaluate(fallback_script)
        if result:
            print(f"✅ Fallback method found element:")
            print(f"  📍 Position: ({result['x']:.1f}, {result['y']:.1f})")
            print(f"  📏 Size: {result['width']:.1f} x {result['height']:.1f}")
            print(f"  🎯 Center: ({result['center_x']:.1f}, {result['center_y']:.1f})")
            return result
        else:
            print(f"❌ Fallback method also failed for backend_node_id: {backend_node_id}")
            return None
            
    except Exception as e:
        print(f"❌ Fallback method error: {e}")
        return None

async def run_highlighting(df_row):
    """
    Function to run the highlighting process
    """
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)  # Use headless=False to see the browser
        page = await browser.new_page()
        
        try:
            # Load the HTML content
            await page.set_content(df_row['raw_html'])
            await page.screenshot(path="original_page.png", full_page=True)
            print("📄 Page loaded and original screenshot saved")
            
            # Parse candidates from their string representation
            pos_candidates = []
            neg_candidates = []
            
            # Handle pos_candidates
            if 'pos_candidates' in df_row and df_row['pos_candidates']:
                try:
                    if isinstance(df_row['pos_candidates'], list):
                        pos_candidates = [json.loads(c) if isinstance(c, str) else c for c in df_row['pos_candidates']]
                    else:
                        pos_candidates = [json.loads(df_row['pos_candidates'])]
                except Exception as e:
                    print(f"Error parsing pos_candidates: {e}")
            
            # Handle neg_candidates  
            if 'neg_candidates' in df_row and df_row['neg_candidates']:
                try:
                    if isinstance(df_row['neg_candidates'], list):
                        neg_candidates = [json.loads(c) if isinstance(c, str) else c for c in df_row['neg_candidates']]
                    else:
                        neg_candidates = [json.loads(df_row['neg_candidates'])]
                except Exception as e:
                    print(f"Error parsing neg_candidates: {e}")
            
            print(f"🎯 Processing {len(pos_candidates)} positive and {len(neg_candidates)} negative candidates")
            
            # Store locations for summary
            pos_locations = []
            
            # Highlight positive candidates (red)
            for i, candidate in enumerate(pos_candidates):
                backend_node_id = candidate.get('backend_node_id')
                if backend_node_id:
                    print(f"\n🔴 Processing positive candidate {i+1} with backend_node_id: {backend_node_id}")
                    location = await highlight_element_by_backend_node_id(page, backend_node_id, 'red')
                    if location:
                        pos_locations.append(location)
                if i == 0:  # Only process first candidate as per your original code
                    break
            
            # Highlight negative candidates (blue) - commented out in your original
            # for i, candidate in enumerate(neg_candidates):
            #     backend_node_id = candidate.get('backend_node_id')
            #     if backend_node_id:
            #         print(f"🔵 Processing negative candidate {i+1} with backend_node_id: {backend_node_id}")
            #         await highlight_element_by_backend_node_id(page, backend_node_id, 'blue')
            
            # Wait for highlights to render
            await page.wait_for_timeout(1000)
            
            # Take final screenshot
            await page.screenshot(path="highlighted_page.png", full_page=True)
            print("📸 Final screenshot saved as highlighted_page.png")
            
            # Print summary
            if pos_locations:
                print(f"\n📊 Summary of {len(pos_locations)} positive candidate locations:")
                for i, loc in enumerate(pos_locations):
                    print(f"  Candidate {i+1}: Center at ({loc['center_x']:.1f}, {loc['center_y']:.1f})")
            
        except Exception as e:
            print(f"Error during execution: {e}")
            import traceback
            traceback.print_exc()
        finally:
            await browser.close()

# Wrapper function that can be called from Jupyter
def highlight_elements(df, row_index=0):
    """
    Convenience function to run the highlighting from Jupyter
    
    Args:
        df: DataFrame containing the HTML and candidates
        row_index: Which row to process (default: 0)
    """
    if row_index >= len(df):
        print(f"Error: row_index {row_index} is out of range. DataFrame has {len(df)} rows.")
        return
    
    # Run the async function
    asyncio.run(run_highlighting(df.iloc[row_index]))

# Example usage:
highlight_elements(df, 0)  # Process first row
```
